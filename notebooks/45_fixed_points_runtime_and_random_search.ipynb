# Fixed points and basins, the hard–flat limit, the coincident fixed point, run time, and tuning by random search

**Computes:** (A) On 1-D and 2-D Gaussian data with one class 20 times larger than the others ($C = 3, 5$): 1,000 k-means++ and 1,000 uniform initializations of FCM, T1d, IT2, T1f and EXP at a default and at the ARI-selected setting; the distinct fixed points, their basins and ARI, and runs that cycle or do not converge. (B) The center shift of one Karnik–Mendel and one Nie–Tan update against the limits of the Remark on the hard–flat limit. (C) The spectral radius of the numerical Jacobian of the update at the coincident fixed point, FCM and IT2 over their settings. (D) Run time per iteration and peak memory on Covertype up to all 581,012 points. (E) Every method tuned over 50 settings drawn at random from continuous ranges, in the design of the paper on the six regime sets. (F) A check of the implementation: the prototype updates divide by a cluster's total weight floored at $10^{-12}$; every setting with a weight exponent of 10 or more is rerun with each cluster's weights rescaled by their largest value (the same prototypes in exact arithmetic) on the stored instances of notebooks 37, 37b and 40, and the comparisons of IT2 are recomputed.

**Inputs:** UCI data; the checkpoints of notebooks 37, 37b and 40 (part F). **Outputs:** `results/45/results_45.json`, checkpoints `runs45*.pkl`, a text report.
**Options:** Runs in parallel over `N_JOBS` processes. Subsets: `N_INIT45`, `CONF45`, `FAMS45`, `SETS45C`, `SIZES45`, `SETS45E`, `N_INST45`, `B45`, `SETS45F`, `SETS45F40`. Saves after every batch and resumes.

```python
import numpy as np
from itertools import permutations

EPS = 1e-12

def fcm_membership(D, m):
    p = 2.0 / (m - 1.0)
    z = -p * np.log(np.maximum(D, EPS))
    z -= z.max(axis=1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(axis=1, keepdims=True)

def hr_bounds(D, m1, m2):
    C = D.shape[1]
    U1, U2 = fcm_membership(D, m1), fcm_membership(D, m2)
    s = 1.0 / (D[:, :, None] / np.maximum(D[:, None, :], EPS)).sum(axis=2)
    cond = s < 1.0 / C
    upper = np.where(cond, U1, U2)
    lower = np.where(cond, U2, U1)
    return lower, upper

def km_interval_1d(x, wl, wu):
    o = np.argsort(x, kind="stable"); xs = x[o]; l = wl[o]; u = wu[o]
    n = len(x)
    cl, cu = np.cumsum(l), np.cumsum(u)
    clx, cux = np.cumsum(l * xs), np.cumsum(u * xs)
    Tl, Tu, Tlx, Tux = cl[-1], cu[-1], clx[-1], cux[-1]
    k = np.arange(n + 1)
    Lk = np.concatenate([[0.0], cl]); Uk = np.concatenate([[0.0], cu])
    Lxk = np.concatenate([[0.0], clx]); Uxk = np.concatenate([[0.0], cux])
    vR_all = (Lxk + (Tux - Uxk)) / np.maximum(Lk + (Tu - Uk), EPS)
    vL_all = (Uxk + (Tlx - Lxk)) / np.maximum(Uk + (Tl - Lk), EPS)
    kR, kL = int(np.argmax(vR_all)), int(np.argmin(vL_all))
    selR = np.zeros(n, bool); selL = np.zeros(n, bool)
    rank = np.empty(n, int); rank[o] = np.arange(n)
    selR[:] = rank >= kR
    selL[:] = rank < kL
    return float(vL_all[kL]), float(vR_all[kR]), selL, selR

def it2fcm_update(X, lower, upper, mc):
    N, M = X.shape; C = lower.shape[1]
    V = np.zeros((C, M)); uL = np.zeros((N, C)); uR = np.zeros((N, C))
    wl, wu = lower ** mc, upper ** mc
    for j in range(C):
        accL = np.zeros(N); accR = np.zeros(N)
        for l in range(M):
            vL, vR, selL, selR = km_interval_1d(X[:, l], wl[:, j], wu[:, j])
            V[j, l] = 0.5 * (vL + vR)
            accL += np.where(selL, upper[:, j], lower[:, j])
            accR += np.where(selR, upper[:, j], lower[:, j])
        uL[:, j], uR[:, j] = accL / M, accR / M
    return V, uL, uR

def dist(X, V):
    return np.maximum(np.linalg.norm(X[:, None, :] - V[None, :, :], axis=2), EPS)

def run_fcm(X, V0, m, iters=300, tol=1e-7):
    V = V0.copy()
    for _ in range(iters):
        U = fcm_membership(dist(X, V), m); W = U ** m
        Vn = (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < tol:
            V = Vn; break
        V = Vn
    U = fcm_membership(dist(X, V), m)
    return V, U.argmax(1)

def run_it2fcm(X, V0, m1, m2, mc=None, iters=300, tol=1e-7):
    mc = 0.5 * (m1 + m2) if mc is None else mc
    V = V0.copy()
    for _ in range(iters):
        lo, up = hr_bounds(dist(X, V), m1, m2)
        Vn, uL, uR = it2fcm_update(X, lo, up, mc)
        if np.abs(Vn - V).max() < tol:
            V = Vn; break
        V = Vn
    lo, up = hr_bounds(dist(X, V), m1, m2)
    _, uL, uR = it2fcm_update(X, lo, up, mc)
    hard_hr = (0.5 * (uL + uR)).argmax(1)
    hard_nearest = dist(X, V).argmin(1)
    return V, hard_hr, hard_nearest

def accuracy(labels, pred, C):
    best = 0.0
    for perm in permutations(range(C)):
        best = max(best, float(np.mean(np.array(perm)[pred] == labels)))
    return best

def centre_error(V, centres):
    C = len(centres); best = np.inf
    for perm in permutations(range(C)):
        best = min(best, float(np.mean(np.linalg.norm(V[list(perm)] - centres, axis=1))))
    return best

def init_pp(X, C, rng):
    idx = [rng.integers(len(X))]
    for _ in range(1, C):
        d2 = np.min(((X[:, None, :] - X[idx][None]) ** 2).sum(2), axis=1)
        idx.append(rng.choice(len(X), p=d2 / d2.sum()))
    return X[idx].copy()

def scenario(name, seed):
    r = np.random.default_rng(seed)
    if name == "equal":
        c = np.array([[0.0, 0.0], [4.0, 0.0]]); n = (100, 100); s = (1.0, 1.0)
        X = np.vstack([r.normal(c[k], s[k], size=(n[k], 2)) for k in range(2)])
    elif name == "volume":
        c = np.array([[0.0, 0.0], [5.0, 0.0]]); n = (100, 100); half = (2.5, 0.6)
        X = np.vstack([c[k] + r.uniform(-half[k], half[k], size=(n[k], 2)) for k in range(2)])
    elif name == "noise":
        c = np.array([[0.0, 0.0], [0.0, 4.0]]); n = (100, 100)
        X = np.vstack([r.normal(c[0], 1.0, size=(n[0], 2)), r.normal(c[1], 1.0, size=(n[1], 2)),
                       r.normal(c[0], 3.0, size=(40, 2))])
        y = np.r_[np.zeros(n[0]), np.ones(n[1]), np.zeros(40)].astype(int)
        return X, y, c, 2
    elif name == "density":
        c = np.array([[0.0, 0.0], [3.5, 0.0]]); n = (200, 30); s = (1.0, 1.0)
        X = np.vstack([r.normal(c[k], s[k], size=(n[k], 2)) for k in range(2)])
    elif name == "volume3":
        c = np.array([[0.0, 0.0], [5.0, 0.0], [2.5, 4.0]]); n = (100, 100, 100); s = (1.4, 0.4, 0.8)
        X = np.vstack([r.normal(c[k], s[k], size=(n[k], 2)) for k in range(3)])
    else:
        raise ValueError(name)
    y = np.concatenate([np.full(n[k], k) for k in range(len(n))])
    return X, y, c, len(n)

from scipy.optimize import linear_sum_assignment
def accuracy_h(labels, pred, C):
    M = np.zeros((C, C))
    np.add.at(M, (labels, pred), 1)
    r, c = linear_sum_assignment(-M)
    return float(M[r, c].sum() / len(labels))

def align(V, ref):
    Dm = np.linalg.norm(V[:, None] - ref[None], axis=2)
    r, c = linear_sum_assignment(Dm)
    out = np.zeros_like(ref); out[c] = V[r]
    return out

def centre_error_h(V, centres):
    return float(np.mean(np.linalg.norm(align(V, centres) - centres, axis=1)))

def real_dataset(name):
    from sklearn import datasets
    d = {"iris": datasets.load_iris, "wine": datasets.load_wine, "breast_cancer": datasets.load_breast_cancer}[name]()
    X = d.data.astype(float); X = (X - X.mean(0)) / X.std(0)
    y = d.target.astype(int)
    return X, y, len(np.unique(y))

def run_fcm_decoupled(X, V0, ma, mb, iters=300, tol=1e-7):
    V = V0.copy()
    for _ in range(iters):
        U = fcm_membership(dist(X, V), ma); W = U ** mb
        Vn = (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < tol:
            V = Vn; break
        V = Vn
    return V, dist(X, V).argmin(1)

def it2fcm_update_fast(X, order, lower, upper, mc):
    N, M = X.shape; C = lower.shape[1]
    Xs = np.take_along_axis(X, order, axis=0)
    rank = np.empty_like(order); np.put_along_axis(rank, order, np.arange(N)[:, None].repeat(M, 1), axis=0)
    V = np.zeros((C, M)); uL = np.zeros((N, C)); uR = np.zeros((N, C))
    z = np.zeros((1, M))
    for j in range(C):
        wl = (lower[:, j] ** mc)[order]; wu = (upper[:, j] ** mc)[order]
        cl, cu = np.cumsum(wl, 0), np.cumsum(wu, 0)
        clx, cux = np.cumsum(wl * Xs, 0), np.cumsum(wu * Xs, 0)
        Lk, Uk = np.vstack([z, cl]), np.vstack([z, cu]); Lxk, Uxk = np.vstack([z, clx]), np.vstack([z, cux])
        Tl, Tu, Tlx, Tux = cl[-1], cu[-1], clx[-1], cux[-1]
        vR_all = (Lxk + (Tux - Uxk)) / np.maximum(Lk + (Tu - Uk), EPS)
        vL_all = (Uxk + (Tlx - Lxk)) / np.maximum(Uk + (Tl - Lk), EPS)
        kR, kL = vR_all.argmax(0), vL_all.argmin(0)
        vR = vR_all[kR, np.arange(M)]; vL = vL_all[kL, np.arange(M)]
        V[j] = 0.5 * (vL + vR)
        selR = rank >= kR[None, :]; selL = rank < kL[None, :]
        up, lo = upper[:, j][:, None], lower[:, j][:, None]
        uR[:, j] = np.where(selR, up, lo).mean(1); uL[:, j] = np.where(selL, up, lo).mean(1)
    return V, uL, uR

def run_it2fcm_fast(X, V0, m1, m2, mc=None, iters=300, tol=1e-7, order=None):
    mc = 0.5 * (m1 + m2) if mc is None else mc
    order = np.argsort(X, axis=0, kind="stable") if order is None else order
    V = V0.copy()
    for _ in range(iters):
        lo, up = hr_bounds(dist(X, V), m1, m2)
        Vn, uL, uR = it2fcm_update_fast(X, order, lo, up, mc)
        if np.abs(Vn - V).max() < tol:
            V = Vn; break
        V = Vn
    lo, up = hr_bounds(dist(X, V), m1, m2)
    _, uL, uR = it2fcm_update_fast(X, order, lo, up, mc)
    return V, (0.5 * (uL + uR)).argmax(1), dist(X, V).argmin(1)

def real_dataset2(name, data_dir="data"):
    import os, pandas as pd
    if name in ("iris", "wine", "wdbc"):
        return real_dataset({"wdbc": "breast_cancer"}.get(name, name))
    if name == "sonar":
        d = pd.read_csv(os.path.join(data_dir, "sonar.csv"), header=None)
        X = d.iloc[:, :-1].to_numpy(float); y = (d.iloc[:, -1] == "M").to_numpy(int)
    elif name == "pima":
        d = pd.read_csv(os.path.join(data_dir, "pima-indians-diabetes.csv"), header=None)
        X = d.iloc[:, :-1].to_numpy(float); y = d.iloc[:, -1].to_numpy(int)
    elif name == "wbc":
        d = pd.read_csv(os.path.join(data_dir, "breast-cancer-wisconsin.data"), header=None, na_values="?").dropna()
        X = d.iloc[:, 1:-1].to_numpy(float); y = (d.iloc[:, -1] == 4).to_numpy(int)
    else:
        raise ValueError(name)
    X = (X - X.mean(0)) / X.std(0)
    return X, y, len(np.unique(y))

EXTRA = {"seeds": "wheat-seeds.csv", "glass": "glass.csv", "ionosphere": "ionosphere.csv",
         "banknote": "banknote_authentication.csv", "ecoli": "ecoli.csv"}
def real_dataset3(name, data_dir="data"):
    import os, pandas as pd
    if name not in EXTRA:
        return real_dataset2(name, data_dir)
    d = pd.read_csv(os.path.join(data_dir, EXTRA[name]), header=None)
    X = d.iloc[:, :-1].to_numpy(float)
    y = pd.factorize(d.iloc[:, -1].astype(str).str.strip(), sort=True)[0]
    X = X[:, X.std(0) > 0]
    X = (X - X.mean(0)) / X.std(0)
    return X, y.astype(int), len(np.unique(y))
```

```python
import time
import numpy as np

def run_it2fcm_nt(X, V0, m1, m2, mc=None, iters=300, tol=1e-6):
    mc = 0.5 * (m1 + m2) if mc is None else mc
    V = V0.copy()
    for _ in range(iters):
        lo, up = hr_bounds(dist(X, V), m1, m2)
        W = 0.5 * (lo ** mc + up ** mc)
        Vn = (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < tol:
            V = Vn; break
        V = Vn
    lo, up = hr_bounds(dist(X, V), m1, m2)
    return V, (0.5 * (lo + up)).argmax(1)

def contingency(y, p, C):
    M = np.zeros((int(y.max()) + 1, C)); np.add.at(M, (y, p), 1); return M

def ari_from_cont(M):
    comb = lambda x: x * (x - 1) / 2.0
    n = M.sum(); s_ij = comb(M).sum(); a = comb(M.sum(1)).sum(); b = comb(M.sum(0)).sum()
    exp = a * b / comb(n); mx = 0.5 * (a + b)
    return float((s_ij - exp) / (mx - exp)) if mx != exp else 1.0

def nmi_from_cont(M):
    n = M.sum(); P = M / n; pi = P.sum(1); pj = P.sum(0)
    nz = P > 0
    I = float((P[nz] * np.log(P[nz] / (pi[:, None] * pj[None, :])[nz])).sum())
    H = lambda q: float(-(q[q > 0] * np.log(q[q > 0])).sum())
    d = 0.5 * (H(pi) + H(pj))
    return I / d if d > 0 else 1.0

def acc_from_cont(M):
    from scipy.optimize import linear_sum_assignment
    r, c = linear_sum_assignment(-M); return float(M[r, c].sum() / M.sum())

def scores(y, p, C):
    M = contingency(y, p, C)
    return dict(acc=acc_from_cont(M), ari=ari_from_cont(M), nmi=nmi_from_cont(M))

def silhouette_pre(Dsub, lab):
    ks = np.unique(lab)
    if len(ks) < 2: return -1.0
    n = len(lab); a = np.zeros(n); b = np.full(n, np.inf)
    for k in ks:
        mk = lab == k; cnt = mk.sum()
        s = Dsub[:, mk].sum(1)
        own = lab == k
        a[own] = s[own] / max(cnt - 1, 1)
        other = ~own
        b[other] = np.minimum(b[other], s[other] / cnt)
    sing = np.array([(lab == lab[i]).sum() == 1 for i in range(n)])
    sil = np.where(sing, 0.0, (b - a) / np.maximum(np.maximum(a, b), EPS))
    return float(sil.mean())

def calinski(X, lab):
    ks = np.unique(lab); n, k = len(X), len(ks)
    if k < 2 or k >= n: return -1.0
    mu = X.mean(0); B = W = 0.0
    for c in ks:
        Xc = X[lab == c]; mc_ = Xc.mean(0)
        B += len(Xc) * ((mc_ - mu) ** 2).sum(); W += ((Xc - mc_) ** 2).sum()
    return float((B / (k - 1)) / (W / (n - k))) if W > 0 else -1.0

def noise_instance(frac, seed):
    r = np.random.default_rng(seed); c = np.array([[0.0, 0.0], [0.0, 4.0]]); k = int(round(frac * 100))
    X = np.vstack([r.normal(c[0], 1.0, (100, 2)), r.normal(c[1], 1.0, (100, 2)), r.normal(c[0], 3.0, (k, 2))])
    y = np.r_[np.zeros(100), np.ones(100), np.zeros(k)].astype(int)
    return X, y, 2

def imbalance_instance(C, ratio, seed):
    r = np.random.default_rng(seed)
    R = 3.5 / (2 * np.sin(np.pi / C)); ang = 2 * np.pi * np.arange(C) / C
    c = np.c_[R * np.cos(ang), R * np.sin(ang)]
    n = np.maximum(np.round(100 * ratio ** (-np.arange(C) / max(C - 1, 1))).astype(int), 3)
    X = np.vstack([r.normal(c[k], 1.0, (n[k], 2)) for k in range(C)])
    y = np.concatenate([np.full(n[k], k) for k in range(C)])
    return X, y, C
```

```python
def hr_bounds_minmax(D, m1, m2):
    U1, U2 = fcm_membership(D, m1), fcm_membership(D, m2)
    return np.minimum(U1, U2), np.maximum(U1, U2)

def run_it2fcm_mm(X, V0, m1, m2, mc=None, iters=300, tol=1e-6, order=None):
    mc = 0.5 * (m1 + m2) if mc is None else mc
    order = np.argsort(X, axis=0, kind="stable") if order is None else order
    V = V0.copy()
    for _ in range(iters):
        lo, up = hr_bounds_minmax(dist(X, V), m1, m2)
        Vn, uL, uR = it2fcm_update_fast(X, order, lo, up, mc)
        if np.abs(Vn - V).max() < tol:
            V = Vn; break
        V = Vn
    lo, up = hr_bounds_minmax(dist(X, V), m1, m2)
    _, uL, uR = it2fcm_update_fast(X, order, lo, up, mc)
    return V, (0.5 * (uL + uR)).argmax(1), dist(X, V).argmin(1)

MM = "IT2FCM-mm"

def run_instance07(X, y, C, s, evaluate=None):
    order = np.argsort(X, axis=0, kind="stable")
    r = np.random.default_rng(8000 + s); sub = np.sort(r.choice(len(X), min(N_SIL, len(X)), replace=False))
    Dsub = np.linalg.norm(X[sub][:, None] - X[sub][None], axis=2)
    rec = {k: dict(ari=[], acc=[], nmi=[], f=[], sil=[], ch=[]) for k in IT2_KEYS}
    t0 = time.perf_counter(); mism_pts = mism_runs = 0
    for q in range(N_INIT):
        V0 = init_pp(X, C, np.random.default_rng(10_000 * (q + 1) + s))
        for k in IT2_KEYS:
            a, b, mode = k; mc = {"m1": a, "mid": 0.5 * (a + b), "m2": b}[mode]
            V, hp, hn = run_it2fcm_mm(X, V0, a, b, mc=mc, tol=TOL, order=order)
            sc = evaluate(V, hp) if evaluate else scores(y, hp, C)
            d = rec[k]
            for m in ("ari", "acc", "nmi"): d[m].append(sc[m])
            d["f"].append(sc.get("f", np.nan)); d["sil"].append(silhouette_pre(Dsub, hp[sub])); d["ch"].append(calinski(X, hp))
            mm_ = int((hp != hn).sum()); mism_pts += mm_; mism_runs += int(mm_ > 0)
    return dict(rec={MM: rec}, sec_per_run={MM: (time.perf_counter() - t0) / (N_INIT * len(IT2_KEYS))},
                mism_points=mism_pts, mism_runs=mism_runs)

def dibco_instance07(i, f, g):
    I8 = np.array(Image.open(f).convert("L")); G = np.array(Image.open(g).convert("L")) < 128
    hist_ink = np.bincount(I8[G].ravel(), minlength=256).astype(float)
    hist_bg = np.bincount(I8[~G].ravel(), minlength=256).astype(float)
    r = np.random.default_rng(5000 + i)
    X = (I8.ravel()[r.choice(I8.size, min(2000, I8.size), replace=False)] / 255.0)[:, None]
    def evaluate(V, lab):
        t = float(V.ravel().mean()); k = int(np.clip(np.ceil(t * 255.0), 0, 256))
        tp, fp = hist_ink[:k].sum(), hist_bg[:k].sum(); fn, tn = hist_ink.sum() - tp, hist_bg.sum() - fp
        M = np.array([[tn, fp], [fn, tp]])
        return dict(acc=float((tp + tn) / M.sum()), ari=ari_from_cont(M), nmi=nmi_from_cont(M), f=float(2 * tp / max(2 * tp + fp + fn, 1)))
    out = run_instance07(X, np.zeros(len(X), int), 2, i, evaluate=evaluate)
    out["name"] = os.path.basename(f).replace("TRAIN_", "").replace("_source", "")
    return out
```

```python
from scipy import stats
MAIN = SYN + REAL + ("dibco",)
SWEEPS = tuple(NOISE) + tuple(IMB)
FAM5 = dict(FAMILIES); FAM5[MM] = IT2_KEYS
PAIRS5 = (("IT2FCM", "FCM"), ("IT2FCM", "T1dec"),
          (MM, "FCM"), (MM, "T1dec"), (MM, "IT2FCM"), (MM, "IT2FCM-NT"))

def halves(nm, R):
    ids = sorted(R); h = len(ids) // 2
    if nm != "dibco": return ids[:h], ids[h:]
    rng = np.random.default_rng(7); years = {i: R[i]["name"][5:9] for i in ids}; fold = {}
    for yv in sorted(set(years.values())):
        ix = [i for i in ids if years[i] == yv]; ix = [ix[j] for j in rng.permutation(len(ix))]
        for j, i in enumerate(ix): fold[i] = j % 2
    return [i for i in ids if fold[i] == 0], [i for i in ids if fold[i] == 1]

def tp(d):
    d = np.asarray(d, float)
    if len(d) < 2 or np.allclose(d, 0): return dict(mean=float(d.mean()) if len(d) else 0.0, t_p=1.0, w_p=1.0, n=len(d))
    t = float(stats.ttest_1samp(d, 0).pvalue)
    try: w = float(stats.wilcoxon(d).pvalue)
    except ValueError: w = 1.0
    return dict(mean=float(d.mean()), t_p=t if np.isfinite(t) else 1.0, w_p=w if np.isfinite(w) else 1.0, n=len(d))

def verdict(a, b):
    sig = lambda z: z["t_p"] < 0.05 and z["w_p"] < 0.05
    if sig(a) and sig(b) and a["mean"] > 0 and b["mean"] > 0: return "BETTER"
    if sig(a) and sig(b) and a["mean"] < 0 and b["mean"] < 0: return "WORSE"
    return "n.s."

def unsup_value(rec_f, metric, by):
    keys = list(rec_f); nq = len(rec_f[keys[0]][by]); vals = []
    for q in range(nq):
        k = max(keys, key=lambda kk: rec_f[kk][by][q]); vals.append(rec_f[k][metric][q])
    return float(np.mean(vals))

def merge(R06, R07):
    ids = sorted(set(R06) & set(R07)); R = {}
    for i in ids:
        R[i] = dict(R06[i]); R[i]["rec"] = dict(R06[i]["rec"]); R[i]["rec"][MM] = R07[i]["rec"][MM]
        R[i]["sec_per_run"] = dict(R06[i]["sec_per_run"]); R[i]["sec_per_run"][MM] = R07[i]["sec_per_run"][MM]
        R[i]["mism_points"] = R07[i].get("mism_points", 0); R[i]["mism_runs"] = R07[i].get("mism_runs", 0)
    return R

def analyse_dataset(nm, R, metric):
    A, B = halves(nm, R); out = {"n": len(R)}
    for by in ("sil", "ch"):
        val = {f: {i: unsup_value(R[i]["rec"][f], metric, by) for i in R} for f in FAM5}
        cmp = {}
        for (x, z) in PAIRS5:
            a = tp([val[x][i] - val[z][i] for i in A]); b = tp([val[x][i] - val[z][i] for i in B])
            cmp[f"{x} vs {z}"] = dict(A=a, B=b, verdict=verdict(a, b), mean=float(np.mean([val[x][i] - val[z][i] for i in R])))
        out[by] = dict(mean={f: float(np.mean(list(val[f].values()))) for f in FAM5}, cmp=cmp)
    lt = {}
    for sn, (tr, te) in (("A->B", (A, B)), ("B->A", (B, A))):
        sel = {f: max(FAM5[f], key=lambda k: np.mean([np.mean(R[i]["rec"][f][k][metric]) for i in tr])) for f in FAM5}
        test = {f: {i: float(np.mean(R[i]["rec"][f][sel[f]][metric])) for i in te} for f in FAM5}
        lt[sn] = dict(selected={f: str(sel[f]) for f in sel}, cmp={f"{x} vs {z}": tp([test[x][i] - test[z][i] for i in te]) for (x, z) in PAIRS5})
    out["label_tuned"] = dict(splits=lt, verdict={c: verdict(lt["A->B"]["cmp"][c], lt["B->A"]["cmp"][c]) for c in lt["A->B"]["cmp"]},
                              mean={c: float(np.mean([lt[s]["cmp"][c]["mean"] for s in lt])) for c in lt["A->B"]["cmp"]})
    out["sec_per_run"] = {f: float(np.median([R[i]["sec_per_run"][f] for i in R])) for f in FAM5}
    out["mm_partition_vs_nearest"] = dict(points=int(sum(R[i]["mism_points"] for i in R)), runs=int(sum(R[i]["mism_runs"] for i in R)))
    out["mm_fingerprint"] = float(np.mean([np.mean(R[i]["rec"][MM][k]["acc"]) for i in R for k in IT2_KEYS]))
    return out

def dataset_level(RES, names, key):
    rows = [n for n in names if n in RES]; res = {}
    for (x, z) in PAIRS5:
        c = f"{x} vs {z}"
        d = np.array([RES[n]["label_tuned"]["mean"][c] if key == "label_tuned" else RES[n][key]["cmp"][c]["mean"] for n in rows])
        w = float(stats.wilcoxon(d).pvalue) if np.any(np.abs(d) > 1e-12) else 1.0
        res[c] = dict(n=len(rows), median=float(np.median(d)), pos=int((d > 0).sum()), neg=int((d < 0).sum()), wilcoxon_p=w)
    if key != "label_tuned":
        M = np.array([[RES[n][key]["mean"][f] for f in FAM5] for n in rows])
        fr = stats.friedmanchisquare(*M.T); ranks = np.mean([stats.rankdata(-r) for r in M], axis=0)
        res["friedman"] = dict(p=float(fr.pvalue), mean_rank=dict(zip(FAM5, map(float, ranks))))
    return res
```

```python
MAXIT = 300
def fcm10(X, V0, m):
    V = V0.copy(); conv = False
    for it in range(1, MAXIT + 1):
        U = fcm_membership(dist(X, V), m); W = U ** m
        Vn = (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < TOL:
            V = Vn; conv = True; break
        V = Vn
    return V, fcm_membership(dist(X, V), m).argmax(1), None, it, conv

def t1d10(X, V0, ma, mb):
    V = V0.copy(); conv = False
    for it in range(1, MAXIT + 1):
        U = fcm_membership(dist(X, V), ma); W = U ** mb
        Vn = (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < TOL:
            V = Vn; conv = True; break
        V = Vn
    return V, dist(X, V).argmin(1), None, it, conv

def it2km10(X, V0, m1, m2, mc, order, bounds):
    V = V0.copy(); conv = False
    for it in range(1, MAXIT + 1):
        lo, up = bounds(dist(X, V), m1, m2)
        Vn, uL, uR = it2fcm_update_fast(X, order, lo, up, mc)
        if np.abs(Vn - V).max() < TOL:
            V = Vn; conv = True; break
        V = Vn
    lo, up = bounds(dist(X, V), m1, m2)
    _, uL, uR = it2fcm_update_fast(X, order, lo, up, mc)
    return V, (0.5 * (uL + uR)).argmax(1), dist(X, V).argmin(1), it, conv

def nt10(X, V0, m1, m2, mc):
    V = V0.copy(); conv = False
    for it in range(1, MAXIT + 1):
        lo, up = hr_bounds(dist(X, V), m1, m2)
        W = 0.5 * (lo ** mc + up ** mc)
        Vn = (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < TOL:
            V = Vn; conv = True; break
        V = Vn
    lo, up = hr_bounds(dist(X, V), m1, m2)
    return V, (0.5 * (lo + up)).argmax(1), None, it, conv

def fit10(fam, key, X, V0, order):
    if fam == "FCM": return fcm10(X, V0, key)
    if fam == "T1dec": return t1d10(X, V0, key[0], key[1])
    a, b, mode = key; mc = {"m1": a, "mid": 0.5 * (a + b), "m2": b}[mode]
    if fam == "IT2FCM": return it2km10(X, V0, a, b, mc, order, hr_bounds)
    if fam == MM: return it2km10(X, V0, a, b, mc, order, hr_bounds_minmax)
    return nt10(X, V0, a, b, mc)

def xie_beni(X, V, m=2.0):
    D = dist(X, V); U = fcm_membership(D, m)
    sep = min(float(((V[j] - V[k]) ** 2).sum()) for j in range(len(V)) for k in range(j + 1, len(V)))
    return float(((U ** m) * D ** 2).sum() / (len(X) * max(sep, EPS)))

def keys_for(fam, G):
    if fam == "FCM": return list(G)
    if fam == "T1dec": return [(a, b) for a in G for b in G]
    return [(a, b, mode) for i, a in enumerate(G) for b in G[i + 1:] for mode in ("m1", "mid", "m2")]

def run_instance10(X, y, C, s, fams, centres=None):
    order = np.argsort(X, axis=0, kind="stable")
    r = np.random.default_rng(8000 + s); sub = np.sort(r.choice(len(X), min(N_SIL, len(X)), replace=False))
    Dsub = np.linalg.norm(X[sub][:, None] - X[sub][None], axis=2)
    fields = ["ari", "acc", "nmi", "sil", "ch", "negxb", "it", "conv"] + (["negcerr"] if centres is not None else [])
    rec = {f: {k: {fl: [] for fl in fields + (["acc_n", "ari_n"] if f == MM else [])} for k in keys} for f, keys in fams.items()}
    V_store = {f: {k: [] for k in keys} for f, keys in fams.items()}; secs = {f: 0.0 for f in fams}
    for q in range(N_INIT):
        V0 = init_pp(X, C, np.random.default_rng(10_000 * (q + 1) + s))
        for fam, keys in fams.items():
            t0 = time.perf_counter()
            for k in keys:
                V, lab, near, it, conv = fit10(fam, k, X, V0, order)
                sc = scores(y, lab, C); d = rec[fam][k]
                for fl in ("ari", "acc", "nmi"): d[fl].append(sc[fl])
                d["sil"].append(silhouette_pre(Dsub, lab[sub])); d["ch"].append(calinski(X, lab)); d["negxb"].append(-xie_beni(X, V))
                d["it"].append(it); d["conv"].append(bool(conv))
                if centres is not None: d["negcerr"].append(-centre_error_h(V, centres))
                if fam == MM:
                    sn = scores(y, near, C); d["acc_n"].append(sn["acc"]); d["ari_n"].append(sn["ari"])
                V_store[fam][k].append(V.astype(np.float32))
            secs[fam] += time.perf_counter() - t0
    return dict(rec=rec, V=V_store, C=C, sec_per_run={f: secs[f] / (N_INIT * len(fams[f])) for f in fams})
```

```python
# (A) Fixed points and basins: 1-D and 2-D Gaussian classes with one class 20 times larger than each of the others (C
# = 3, 5).
from scipy.optimize import linear_sum_assignment as _lsa
N_INIT45 = int(os.environ.get("N_INIT45", "1000")); TOL45 = float(os.environ.get("TOL45", "0.01"))
FAMS45 = ("FCM", "T1dec", MM, "T1f", "EXP")
if os.environ.get("FAMS45"): FAMS45 = tuple(os.environ["FAMS45"].split(","))
DEFAULT45 = {"FCM": 2.0, "T1dec": (2.0, 2.0), MM: (1.5, 5.0, "mid"), "T1f": (2.0, 2.0, 1.0), "EXP": (2.0, 2.0, 4.0)}
def toy45(dim, C, seed=45, big=2000, ratio=20):
    rng = np.random.default_rng(seed); n = [big] + [big // ratio] * (C - 1)
    if dim == 1: mu = np.array([[0.0]] + [[4.0 + 3.0 * k] for k in range(C - 1)])
    else:
        ang = np.linspace(0, 2 * np.pi, C - 1, endpoint=False); mu = np.vstack([[0.0, 0.0]] + [[4.5 * np.cos(a), 4.5 * np.sin(a)] for a in ang])
    sd = np.array([1.5] + [0.5] * (C - 1))
    X = np.vstack([mu[k] + sd[k] * rng.standard_normal((n[k], dim)) for k in range(C)]); y = np.repeat(np.arange(C), n)
    return X, y
def fit_v(fam, key, X, V0, order):
    """Final prototypes, labels, convergence flag and iterations of one run."""
    if fam in ("FCM", "T1dec", MM): V, lab, _, it, conv = fit10(fam, key, X, V0, order); return V, lab, conv, it
    if fam == "T1f": V, lab, it, conv = t1f10(X, V0, *key); return V, lab, conv, it
    if fam == "EXP": V, lab, it, conv = exp10(X, V0, key[0], key[1], key[2], order); return V, lab, conv, it
    raise KeyError(fam)
def step_v(fam, key, X, V, order):
    """One update of the prototypes (the map whose fixed points the runs reach)."""
    D = dist(X, V)
    if fam == "FCM": W = fcm_membership(D, key) ** key
    elif fam == "T1dec": W = fcm_membership(D, key[0]) ** key[1]
    elif fam == "T1f": W = np.maximum(fcm_membership(D, key[0]) ** key[1], key[2] * len(V) ** (-key[1]))
    elif fam == "EXP": W = fcm_membership(D, key[0]) ** key[1]; return km_center_w(X, order, W, (1.0 + key[2]) * W)
    elif fam == MM:
        a, b = key[0], key[1]; mc = key[2] if not isinstance(key[2], str) else {"m1": a, "mid": 0.5 * (a + b), "m2": b}[key[2]]
        lo, up = hr_bounds_minmax(D, a, b); return it2fcm_update_fast(X, order, lo, up, mc)[0]
    else: raise KeyError(fam)
    return (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
def classify_nc(fam, key, X, V, order, more=2700):
    """A run still moving after 300 iterations: continued up to 3,000; 'slow' if it then converges, '2-cycle' if V_{t+2} = V_t."""
    for _ in range(more):
        Vn = step_v(fam, key, X, V, order)
        if np.abs(Vn - V).max() < TOL: return "slow", Vn
        V = Vn
    b = step_v(fam, key, X, step_v(fam, key, X, V, order), order)
    return ("2-cycle" if np.abs(b - V).max() < 10 * TOL else "no convergence"), V
def canon(V):
    return V[np.lexsort(V.T[::-1])]
def group_fixed(Vs, tol):
    reps, idx = [], []
    for V in Vs:
        for i, R in enumerate(reps):
            r, c = _lsa(dist(V, R));
            if np.abs(V[r] - R[c]).max() < tol: idx.append(i); break
        else: reps.append(V); idx.append(len(reps) - 1)
    return reps, np.array(idx)
RES45A = {}; outA45, pathA45 = ckpt("runs45a_basins.pkl")
CONF45 = [tuple(int(v) for v in x.split(":")) for x in os.environ.get("CONF45", "1:3,1:5,2:3,2:5").split(",") if x]
for dim, C in CONF45:
    if True:
        tag = f"d{dim}_C{C}"; X, y = toy45(dim, C); order = np.argsort(X, axis=0, kind="stable"); rngX = float(np.ptp(X))
        res = {}
        for fam in FAMS45:
            ks = keys_x(fam, G_B)
            if (tag, fam, "tuned") not in outA45:
                tune = lambda k: float(np.mean([_ari(y, fit_v(fam, k, X, init_pp(X, C, np.random.default_rng(70_000 + q)), order)[1]) for q in range(10)]))
                sc = Parallel(n_jobs=N_JOBS)(delayed(tune)(k) for k in ks) if N_JOBS > 1 else [tune(k) for k in ks]
                outA45[(tag, fam, "tuned")] = ks[int(np.argmax(sc))]; save_ckpt(outA45, pathA45)
            for which, key in (("default", DEFAULT45[fam]), ("tuned", outA45[(tag, fam, "tuned")])):
                ck = (tag, fam, which, "runs")
                if ck not in outA45:
                    def one(q, kind):
                        rng = np.random.default_rng(80_000 + q + (0 if kind == "pp" else 1_000_000))
                        V0 = init_pp(X, C, rng) if kind == "pp" else rng.uniform(X.min(0), X.max(0), (C, X.shape[1]))
                        V, lab, conv, _ = fit_v(fam, key, X, V0, order); status = "converged"
                        if not conv: status, V = classify_nc(fam, key, X, V, order); lab = dist(X, V).argmin(1)
                        return kind, V, float(_ari(y, lab)), status, float(score_x(y, lab, C)["kept"]), int(len(np.unique(lab)))
                    runs = Parallel(n_jobs=N_JOBS)(delayed(one)(q, kind) for kind in ("pp", "unif") for q in range(N_INIT45)) if N_JOBS > 1 else [one(q, kind) for kind in ("pp", "unif") for q in range(N_INIT45)]
                    outA45[ck] = runs; save_ckpt(outA45, pathA45)
                runs = outA45[ck]; d = {}
                for kind in ("pp", "unif"):
                    rr = [r for r in runs if r[0] == kind]; reps, idx = group_fixed([canon(r[1]) for r in rr], TOL45 * rngX)
                    basins = sorted(({"share": float((idx == i).mean()), "ari": float(np.mean([rr[j][2] for j in np.where(idx == i)[0]])),
                                      "kept": float(np.mean([rr[j][4] for j in np.where(idx == i)[0]])), "clusters": int(np.median([rr[j][5] for j in np.where(idx == i)[0]]))}
                                     for i in range(len(reps))), key=lambda b: -b["share"])
                    st = {s_: sum(r[3] == s_ for r in rr) for s_ in ("converged", "slow", "2-cycle", "no convergence")}
                    d[kind] = dict(n_fixed=len(reps), basins=basins[:6], best_ari=max(b["ari"] for b in basins), share_best=sum(b["share"] for b in basins if b["ari"] >= max(b2["ari"] for b2 in basins) - 1e-6),
                                   mean_ari=float(np.mean([r[2] for r in rr])), status=st)
                res[f"{NAMEX[fam]} {which} {key}"] = d
        RES45A[tag] = res
        print(f"{tag}: " + "  ".join(f"{k}: {v['pp']['n_fixed']} fixed pts, best-basin {100*v['pp']['share_best']:.0f}%" for k, v in res.items()), flush=True)

RES45B = []
for C in (2, 3, 5):
    for nj, nb, D in ((100, 10, 4.0), (100, 50, 4.0), (40, 120, 3.0), (200, 20, 6.0)):
        X = np.r_[np.zeros(nj), np.full(nb, D)][:, None] + 1e-9 * np.arange(nj + nb)[:, None]; order = np.argsort(X, axis=0, kind="stable")
        V = np.array([[-0.1]] + [[D + 1.5 + 3.0 * k] for k in range(C - 1)])
        for m1, m2 in ((1.05, 20.0), (1.01, 200.0), (1.001, 2000.0)):
            lo, up = hr_bounds_minmax(dist(X, V), m1, m2); Vk, _, _ = it2fcm_update_fast(X, order, lo, up, m1)
            W = 0.5 * (lo ** m1 + up ** m1); Vn = (W.T @ X) / W.sum(0)[:, None]
            RES45B.append(dict(C=C, n_j=nj, n_b=nb, D=D, m1=m1, m2=m2, km_shift=float(Vk[0, 0]), km_limit=nb * D / (2 * (nj + nb)),
                               nt_shift=float(Vn[0, 0]), nt_limit=nb * D / ((C + 1) * nj + nb)))

SETS45C = ["imb_C3_r20", "imb_C5_r20", "imb_C8_r20", "ecoli", "dermatology", "hcv", "iris"]
if os.environ.get("SETS45C") is not None: SETS45C = [x for x in os.environ["SETS45C"].split(",") if x]
def jac_radius(fam, key, X, C, h=1e-5, seed=0):
    M = X.shape[1]; Vs = np.repeat(X.mean(0)[None], C, 0); order = np.argsort(X, axis=0, kind="stable")
    rng = np.random.default_rng(seed); Vs = Vs + 1e-7 * rng.standard_normal(Vs.shape)
    J = np.zeros((C * M, C * M))
    for i in range(C * M):
        E = np.zeros(C * M); E[i] = h; E = E.reshape(C, M)
        J[:, i] = ((step_v(fam, key, X, Vs + E, order) - step_v(fam, key, X, Vs - E, order)) / (2 * h)).ravel()
    return float(np.abs(np.linalg.eigvals(J)).max())
RES45C = {}
for nm in SETS45C:
    try: X, y, C = inst_set(nm, 0)
    except Exception as e: print(f"{nm}: data not available ({e!r}), skipped"); continue
    if X.shape[1] * C > 120: X = X[:, :max(1, 120 // C)]
    t0 = time.time(); G = grid_set(nm)
    fk = [("FCM", m) for m in G] + [(MM, k) for k in keys_for(MM, G) if k[2] == "mid"]
    rad = Parallel(n_jobs=N_JOBS)(delayed(jac_radius)(f, k, X, C) for f, k in fk) if N_JOBS > 1 else [jac_radius(f, k, X, C) for f, k in fk]
    RES45C[nm] = dict(M=X.shape[1], C=C, radius={f"{NAMEX[f]} {k}": r for (f, k), r in zip(fk, rad)},
                      unstable_share={NAMEX[f]: float(np.mean([r > 1 + 1e-6 for (ff, k), r in zip(fk, rad) if ff == f])) for f in ("FCM", MM)})
    print(f"{nm:12s} share of settings with an unstable coincident point: " + ", ".join(f"{k} {100*v:.0f}%" for k, v in RES45C[nm]["unstable_share"].items()) + f" ({time.time()-t0:.0f}s)", flush=True)

import tracemalloc
SIZES45 = tuple(int(float(v)) for v in os.environ.get("SIZES45", "1000,10000,100000,581012").split(","))
FAMS45D = ("FCM", "T1dec", MM, "T1f", "EXP", "GMM", "KM")
KEY45D = {"FCM": 2.0, "T1dec": (2.0, 2.0), MM: (1.5, 5.0, "mid"), "T1f": (2.0, 2.0, 1.0), "EXP": (2.0, 2.0, 4.0), "GMM": ("diag", 1e-4), "KM": "km10"}
RES45D = {}
try: Xc, yc, Cc = data35("covertype")
except Exception as e: Xc = None; print(f"covertype: data not available ({e!r}), part D skipped")
if Xc is not None:
    for N in SIZES45:
        N = min(N, len(yc)); idx = np.sort(np.random.default_rng(45).choice(len(yc), N, replace=False)); X = Xc[idx]; order = np.argsort(X, axis=0, kind="stable")
        V0 = init_pp(X, Cc, np.random.default_rng(1)); row = {}
        for fam in FAMS45D:
            old = MAXIT; MAXIT = 50; tracemalloc.start(); t0 = time.perf_counter(); its = None
            try:
                if fam in ("FCM", "T1dec", MM, "T1f", "EXP"): its = fit_v(fam, KEY45D[fam], X, V0, order)[3]
                else: fit_x(fam, KEY45D[fam], X, V0, order, 1)
            finally: MAXIT = old
            dt = time.perf_counter() - t0; peak = tracemalloc.get_traced_memory()[1]; tracemalloc.stop()
            row[NAMEX[fam]] = dict(seconds=dt, iterations=its, per_iteration=(dt / its if its else None), peak_mb=peak / 2 ** 20)
        sort_t = time.perf_counter(); np.argsort(X, axis=0, kind="stable"); row["sorting (once, IT2 and EXP)"] = dict(seconds=time.perf_counter() - sort_t, iterations=None, per_iteration=None, peak_mb=0.0)
        RES45D[N] = row
        print(f"N={N:7d}: " + ", ".join(f"{k} {v['seconds']:.2f}s/{v['peak_mb']:.0f}MB" for k, v in row.items()), flush=True)

B45 = int(os.environ.get("B45", "50"))
def _lu(r, lo, hi, n): return np.exp(r.uniform(np.log(lo), np.log(hi), n))
def rand_keys(fam, B=B45):
    r = np.random.default_rng({"FCM": 1, "T1dec": 2, MM: 3, "T1f": 4, "EXP": 5}[fam] + 4500)
    if fam == "FCM": return [float(round(v, 4)) for v in _lu(r, 1.05, 20, B)]
    a, b = _lu(r, 1.05, 20, B), _lu(r, 1.05, 20, B)
    if fam == "T1dec": return [(float(round(x, 4)), float(round(z, 4))) for x, z in zip(a, b)]
    if fam == MM:
        lo, hi = np.minimum(a, b), np.maximum(a, b); hi = np.where(hi - lo < 0.05, lo + 0.05, hi); mc = r.uniform(lo, hi)
        return [(float(round(x, 4)), float(round(z, 4)), float(round(w, 4))) for x, z, w in zip(lo, hi, mc)]
    if fam == "T1f": return [(float(round(x, 4)), float(round(z, 4)), float(round(e, 4))) for x, z, e in zip(a, b, _lu(r, 0.25, 4, B))]
    if fam == "EXP": return [(float(round(x, 4)), float(round(z, 4)), float(round(c, 4))) for x, z, c in zip(a, b, _lu(r, 0.5, 32, B))]
if "_keys_x45" not in globals(): _keys_x45, _fit_x45 = keys_x, fit_x
def keys_x(fam, G):
    if fam.endswith("~r"): return rand_keys(fam[:-2])
    return _keys_x45(fam, G)
def fit_x(fam, key, X, V0, order, seed):
    if fam.endswith("~r"):
        f = fam[:-2]
        if f == MM: return it2km10(X, V0, key[0], key[1], key[2], order, hr_bounds_minmax)[1]
        return _fit_x45(f, key, X, V0, order, seed)
    return _fit_x45(fam, key, X, V0, order, seed)
for f in ("FCM", "T1dec", MM, "T1f", "EXP"): NAMEX[f + "~r"] = NAMEX[f]
SETS45E = ["imb_C3_r20", "imb_C5_r20", "imb_C8_r20", "ecoli", "dermatology", "hcv"]
if os.environ.get("SETS45E") is not None: SETS45E = [x for x in os.environ["SETS45E"].split(",") if x]
FAMS45E = tuple(f + "~r" for f in ("FCM", "T1dec", MM, "T1f", "EXP"))
N45 = os.environ.get("N_INST45")
RES45E = {}; outE45, pathE45 = ckpt("runs45e_random.pkl")
for nm in SETS45E:
    try: inst_set(nm, 0)
    except Exception as e: print(f"{nm}: data not available ({e!r}), skipped"); continue
    t0 = time.time(); n = min(n_set(nm), int(N45)) if N45 else n_set(nm); G = grid_set(nm)
    run_many_g([((nm, s), (lambda nm=nm, s=s: inst_set(nm, s) + (s,)), FAMS45E, G) for s in range(n)], pathE45, outE45)
    recs = [outE45[(nm, s)]["rec"] for s in range(n)]
    RES45E[nm] = dict(compare={NAMEX[z]: lt_both(recs, MM + "~r", z, nm) for z in FAMS45E if z != MM + "~r"},
                      level={NAMEX[f]: {m: level_at(recs, f, m) for m in ("ari", "bacc", "kept", "tiny")} for f in FAMS45E})
    print(f"{nm:12s} IT2 minus (random search, {B45} settings each): " + "  ".join(f"{z} {100*c['ari']['mean']:+.1f}{mk(c['ari'])}{mkc(c['ari'])}" for z, c in RES45E[nm]["compare"].items()) + f" ({time.time()-t0:.0f}s)", flush=True)

def _logw(U, e): return e * np.log(np.maximum(U, 1e-300))
def _scaled(LW): return np.exp(LW - LW.max(0, keepdims=True))
def fit_scaled(fam, key, X, V0, order):
    V = V0.copy(); C = len(V0)
    for it in range(1, MAXIT + 1):
        D = dist(X, V)
        if fam == MM:
            a, b = key[0], key[1]; mc = {"m1": a, "mid": 0.5 * (a + b), "m2": b}[key[2]]
            lo, up = hr_bounds_minmax(D, a, b); sc = 1.0 / np.maximum(up.max(0, keepdims=True), 1e-300)
            Vn = it2fcm_update_fast(X, order, lo * sc, up * sc, mc)[0]
        else:
            ma, mb = (key, key) if fam == "FCM" else (key[0], key[1]); U = fcm_membership(D, ma); LW = _logw(U, mb)
            if fam == "T1f": LW = np.maximum(LW, np.log(key[2]) - mb * np.log(C))
            W = _scaled(LW)
            Vn = km_center_w(X, order, W, (1.0 + key[2]) * W) if fam == "EXP" else (W.T @ X) / np.maximum(W.sum(0)[:, None], EPS)
        if np.abs(Vn - V).max() < TOL: V = Vn; break
        V = Vn
    if fam == MM:
        lo, up = hr_bounds_minmax(dist(X, V), key[0], key[1]); _, uL, uR = it2fcm_update_fast(X, order, lo, up, {"m1": key[0], "mid": 0.5 * (key[0] + key[1]), "m2": key[1]}[key[2]])
        return (0.5 * (uL + uR)).argmax(1)
    return dist(X, V).argmin(1)
def w_exp(fam, key):
    if fam == "FCM": return key
    if fam == MM: return {"m1": key[0], "mid": 0.5 * (key[0] + key[1]), "m2": key[1]}[key[2]]
    return key[1]
FAMS45F = ("FCM", "T1dec", MM, "T1f", "EXP")
def rerun_scaled(X, y, C, s, fams, G, emin=10.0):
    order = np.argsort(X, axis=0, kind="stable"); out = {}
    for fam in fams:
        out[fam] = {}
        for k in keys_x(fam, G):
            if w_exp(fam, k) < emin: continue
            vals = [score_x(y, fit_scaled(fam, k, X, init_pp(X, C, np.random.default_rng(10_000 * (q + 1) + s)), order), C) for q in range(N_INIT)]
            out[fam][k] = {m: float(np.mean([v[m] for v in vals])) for m in vals[0]}
    return out
def splice(rec, new):
    r = {f: dict(v) for f, v in rec.items()}
    for f, d in new.items():
        if f in r: r[f].update({k: v for k, v in d.items() if k in r[f]})
    return r
RES45F = {}; outF37, pathF37 = ckpt("runs45f_scaled37.pkl"); outF40, pathF40 = ckpt("runs45f_scaled40.pkl")
st37 = {**load_other("37b", "runs37a.pkl"), **load_other("37", "runs37a.pkl")}; st40 = load_other("40", "runs40_disjoint.pkl")
SETS45F = sorted({k[0] for k in st37}) if st37 else []
if os.environ.get("SETS45F") is not None: SETS45F = [x for x in os.environ["SETS45F"].split(",") if x]
def changed(old, new):
    d = [abs(new[f][k]["ari"] - old[f][k]["ari"]) for f in new if f in old for k in new[f] if k in old[f]]
    return dict(settings=len(d), changed=int(sum(x > 0.005 for x in d)), max_change=float(max(d) if d else 0.0))
for nm in SETS45F:
    ss = sorted(s for (n_, s) in st37 if n_ == nm); fams = [f for f in FAMS45F if f in st37[(nm, ss[0])]["rec"]]; G = grid_set(nm)
    todo = [s for s in ss if (nm, s) not in outF37]
    for b in range(0, len(todo), max(1, 2 * N_JOBS)):
        part = todo[b:b + max(1, 2 * N_JOBS)]
        res = Parallel(n_jobs=N_JOBS)(delayed(rerun_scaled)(*inst_set(nm, s), s, fams, G) for s in part) if N_JOBS > 1 else [rerun_scaled(*inst_set(nm, s), s, fams, G) for s in part]
        for s, r_ in zip(part, res): outF37[(nm, s)] = r_
        save_ckpt(outF37, pathF37)
    old = [st37[(nm, s)]["rec"] for s in ss]; new = [splice(st37[(nm, s)]["rec"], outF37[(nm, s)]) for s in ss]
    ch = [changed(st37[(nm, s)]["rec"], outF37[(nm, s)]) for s in ss]
    r = dict(design="paper", settings_rerun=sum(c["settings"] for c in ch), changed=sum(c["changed"] for c in ch), max_change=max(c["max_change"] for c in ch), compare={})
    for z in fams:
        if z == MM: continue
        a, b_ = lt_both(old, MM, z, nm), lt_both(new, MM, z, nm)
        r["compare"][NAMEX[z]] = dict(old=a["ari"]["mean"], new=b_["ari"]["mean"], old_verdict=a["ari"]["verdict"], new_verdict=b_["ari"]["verdict"],
                                      old_selected=a["selected"], new_selected=b_["selected"])
    RES45F[f"{nm} (design of the paper)"] = r
    print(f"{nm:12s} paper design: {r['changed']}/{r['settings_rerun']} rescaled settings change ARI (max {100*r['max_change']:.1f}); IT2 minus " + "  ".join(f"{z} {100*c['old']:+.1f}->{100*c['new']:+.1f}" for z, c in r["compare"].items()), flush=True)
SETS45F40 = [nm for nm in REGIME_REAL if any(k[0] == nm for k in st40)] if st40 else []
if os.environ.get("SETS45F40") is not None: SETS45F40 = [x for x in os.environ["SETS45F40"].split(",") if x]
for nm in SETS45F40:
    try: X_all, y_all, C = full_any(nm); X_all = np.asarray(X_all, float); y_all = np.asarray(y_all, int)
    except Exception as e: print(f"{nm}: data not available ({e!r}), skipped"); continue
    fams = [f for f in FAMS45F if f in st40[(nm, 0, 0, 0)]["rec"]]; G = grid_set(nm)
    jobs, fh = disjoint_jobs(nm, X_all, y_all, C, fams, G); todo = [j for j in jobs if j[0] not in outF40]
    for b in range(0, len(todo), max(1, 2 * N_JOBS)):
        part = todo[b:b + max(1, 2 * N_JOBS)]; inst = [j[1]() for j in part]
        res = Parallel(n_jobs=N_JOBS)(delayed(rerun_scaled)(*i_, fams, G) for i_ in inst) if N_JOBS > 1 else [rerun_scaled(*i_, fams, G) for i_ in inst]
        for j, r_ in zip(part, res): outF40[j[0]] = r_
        save_ckpt(outF40, pathF40)
    keys = [j[0] for j in jobs]; newd = {k: dict(rec=splice(st40[k]["rec"], outF40[k])) for k in keys}
    ch = [changed(st40[k]["rec"], outF40[k]) for k in keys]
    r = dict(design="point-disjoint", settings_rerun=sum(c["settings"] for c in ch), changed=sum(c["changed"] for c in ch), max_change=max(c["max_change"] for c in ch), compare={})
    vo, vn = disjoint_values(st40, nm, fams), disjoint_values(newd, nm, fams)
    for z in fams:
        if z == MM: continue
        a, b_ = disjoint_compare(vo, MM, z), disjoint_compare(vn, MM, z)
        r["compare"][NAMEX[z]] = dict(old=a["mean"], new=b_["mean"], old_p=a["p_5x2F"], new_p=b_["p_5x2F"])
    RES45F[f"{nm} (point-disjoint)"] = r
    print(f"{nm:12s} point-disjoint: {r['changed']}/{r['settings_rerun']} rescaled settings change ARI (max {100*r['max_change']:.1f}); IT2 minus " + "  ".join(f"{z} {100*c['old']:+.1f}->{100*c['new']:+.1f}" for z, c in r["compare"].items()), flush=True)
```
